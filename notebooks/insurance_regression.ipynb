```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression
```

The data file to be read is insurance.csv . 

```python
our_data=pd.read_csv("insurance.csv")
our_data.head(10)
```

Age Column and Charges Column from the Insurance Table are at location 0 and -1 or last column...

```python
# The iloc function allows use to select rows from the dataframe for plotting.  

x = our_data.iloc[:,0].values #age column

y = our_data.iloc[:,-1].values #charges column

#print('size of x = {0}'.format(x.size))
#print('size of y = {0}'.format(y.size))


plt.scatter(x,y,color = 'b')
plt.xlabel('Age (Number of Years)')
plt.ylabel('Charges (£)')
plt.show()
```

Plot the data and model...

```python
insurance_model=LinearRegression()

X =our_data.iloc[:,0].values.reshape(-1,1)
#Y=our_data.iloc[:,-1].values.reshape(-1,1)
insurance_model.fit(X,y)

# Plot the data and model
y_pred =insurance_model.predict(X)

plt.scatter(X,y,color = 'b')
plt.plot(X,y_pred,color = 'r')
plt.xlabel('Age (Number of Years)')
plt.ylabel('Charges (£)')
plt.show()

#_____________________End of Task_____________________

# Calculation of coefficient for age 45..
age =45

print(f"Insurance Charges be £ {round(float(insurance_model.predict([[age]])),2)} for a person aged {age}.")
```
